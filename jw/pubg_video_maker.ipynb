```python
from moviepy.editor import ImageSequenceClip
import os

def create_video_from_images(image_folder, output_video_path, fps=15):
    images = [os.path.join(image_folder, img) for img in sorted(os.listdir(image_folder)) if img.endswith(".png")]
    clip = ImageSequenceClip(images, fps=fps)
    clip.write_videofile(output_video_path, codec='libx264')

image_folder = "scatter_plots"
output_video_path = "pubg_video.mp4"
create_video_from_images(image_folder, output_video_path)
```
